## GIT Avanzado

### Creación de Branches

- Los branches son repositorios internos al main que se usan para resolver errores con el código
- Permiten que por ejemplo un sitio web esté en producción, y al mismo tiempo repararlo
- Lo que se hace al generar un branch es COPIAR todos los archivos del main, para solamente resolver algunos, y luego se hace un MERGE que los soluciona
- Para generar el branch se hace con "git create branch" - crear nombre
- Se ve en la esquina inferior izquierda el nombre del branch donde se está trabajando
- Se hace la publicación del branch para que aparezca en github: "git publish branch"
- Se puede validar en github la presencia del branch, luego del publish

### Cambios en el branch y merge

- Se hacen los cambios locales en el archivo
- Se hace un git commit y push para subir el cambio EN LA RAMA recién creada
- Se valida que solo hay el cambio en la rama, no en el main, en Github button synchronize

### Checkout

- También en el VS Code puedo hacer click en la esquina inferior izquierda y cambiar la rama o el main que estoy viendo (checkout). También se puede hacer un "git checkout to - select branch" Button publish branch
